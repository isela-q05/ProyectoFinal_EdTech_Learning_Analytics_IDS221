### Interpretación del campo `historial_cualitativo`

El campo `historial_cualitativo` contiene un **diccionario en formato string** con la bitácora del estudiante. Vamos a extraer información valiosa:
- `total_observaciones`: Cuántas veces ha sido observado.
- `requiere_seguimiento_especial`: Si necesita apoyo adicional.
- `ultima_intervencion`: Último tipo de apoyo recibido.
- `estado_ultimo`: Estado de esa última intervención.

```python
# ==========================================
# INTERPRETACIÓN DEL JSON
# ==========================================

def extraer_info_json(json_str):
    try:
        data = ast.literal_eval(json_str) if isinstance(json_str, str) else json_str
        total_obs = data.get('total_observaciones', 0)
        req_seguimiento = data.get('requiere_seguimiento_especial', False)
        bitacora = data.get('bitacora', [])
        ultima_intervencion = bitacora[-1]['tipo_intervencion'] if bitacora else 'Sin intervención'
        estado_ultimo = bitacora[-1]['estado'] if bitacora else 'Sin estado'
        return pd.Series([total_obs, req_seguimiento, ultima_intervencion, estado_ultimo])
    except:
        return pd.Series([0, False, 'Sin intervención', 'Sin estado'])

df_limpio[['total_observaciones', 'requiere_seguimiento', 'ultima_intervencion', 'estado_ultimo']] = df_limpio['historial_cualitativo'].apply(extraer_info_json)

print("JSON interpretado. Nuevas columnas:")
print(df_limpio[['total_observaciones', 'requiere_seguimiento', 'ultima_intervencion', 'estado_ultimo']].head(5))
```

### Características creadas y su justificación

Se seleccionaron **6 características** que responden directamente a las preguntas hechas:

| # | Característica | Tipo | Pregunta que responde | Regla de negocio |
|---|---------------|------|----------------------|------------------|
| 1 | `promedio_academico` | Numérica | ¿Cuál es el rendimiento promedio del estudiante? | (nota_teoria + nota_laboratorio) / 2 |
| 2 | `porcentaje_asistencia` | Normalizada | ¿Cuál es su nivel de asistencia? | (10 - faltas_consecutivas) / 10 * 100 |
| 3 | `actividad_lms` | Numérica | ¿Qué tan activo es en la plataforma? | tareas_entregadas × nivel_participación |
| 4 | `riesgo_academico` | Categórica | ¿Cuál es su nivel de riesgo académico? | Clasificación: Alto / Medio / Bajo |
| 5 | `es_fin_semana` | Booleana | ¿La intervención fue en fin de semana? | fecha_bitacora.day_of_week >= 5 |
| 6 | `semana_evaluacion` | Temporal | ¿En qué semana del año fue evaluado? | Extraída de la bitácora JSON |

```python
# ==========================================
# INGENIERÍA DE CARACTERÍSTICAS (6 VARIABLES CLAVE)
# ==========================================

# ============================================
# 1. PROMEDIO ACADÉMICO (Numérica)
# ============================================
# Responde: ¿Cuál es el rendimiento promedio del estudiante?
# Regla: Promedio entre nota de teoría y laboratorio

df_limpio['promedio_academico'] = (df_limpio['nota_teoria'] + df_limpio['nota_laboratorio']) / 2

# ============================================
# 2. PORCENTAJE DE ASISTENCIA (Normalizada)
# ============================================
# Responde: ¿Cuál es su nivel de asistencia?
# Regla: Asumimos 10 faltas como el máximo permitido en el ciclo

df_limpio['porcentaje_asistencia'] = ((10 - df_limpio['faltas_consecutivas']) / 10 * 100).clip(lower=0)

# ============================================
# 3. ACTIVIDAD LMS (Numérica)
# ============================================
# Responde: ¿Qué tan activo es el estudiante en la plataforma?
# Regla: Combina tareas entregadas con nivel de participación

mapa_participacion = {'Baja': 1, 'Media': 2, 'Alta': 3}
df_limpio['actividad_lms'] = (
    df_limpio['tareas_entregadas'] *
    df_limpio['participacion_nivel'].map(mapa_participacion).fillna(0)
)

# ============================================
# 4. RIESGO ACADÉMICO (Categórica)
# ============================================
# Responde: ¿Cuál es su nivel de riesgo académico?
# Regla: Clasificación basada en nota final y probabilidad de deserción

def clasificar_riesgo_academico(row):
    if row['nota_final'] < 6.0 or row['probabilidad_desercion'] > 0.7:
        return 'Alto'
    elif row['nota_final'] < 7.0 or row['probabilidad_desercion'] > 0.4:
        return 'Medio'
    else:
        return 'Bajo'

df_limpio['riesgo_academico'] = df_limpio.apply(clasificar_riesgo_academico, axis=1)

# ============================================
# 5. ES_FIN_SEMANA (Booleana)
# ============================================
# Responde: ¿La intervención fue en fin de semana?
# Regla: True si el día de la última intervención fue sábado o domingo

def es_fin_semana(json_str):
    try:
        data = ast.literal_eval(json_str) if isinstance(json_str, str) else json_str
        bitacora = data.get('bitacora', [])
        if bitacora:
            fecha = pd.to_datetime(bitacora[-1]['fecha'])
            return fecha.day_of_week >= 5
        return False
    except:
        return False

df_limpio['es_fin_semana'] = df_limpio['historial_cualitativo'].apply(es_fin_semana)

# ============================================
# 6. SEMANA DE EVALUACIÓN (Temporal)
# ============================================
# Responde: ¿En qué semana del año fue evaluado?
# Regla: Extraer el número de semana ISO de la última intervención

def extraer_semana_evaluacion(json_str):
    try:
        data = ast.literal_eval(json_str) if isinstance(json_str, str) else json_str
        bitacora = data.get('bitacora', [])
        if bitacora:
            fecha = pd.to_datetime(bitacora[-1]['fecha'])
            return fecha.isocalendar()[1]
        return 0
    except:
        return 0

df_limpio['semana_evaluacion'] = df_limpio['historial_cualitativo'].apply(extraer_semana_evaluacion)

# ============================================
# VERIFICACIÓN DE LAS 6 CARACTERÍSTICAS
# ============================================

print("6 Características clave creadas exitosamente:\n")
print(df_limpio[[
    'promedio_academico',
    'porcentaje_asistencia',
    'actividad_lms',
    'riesgo_academico',
    'es_fin_semana',
    'semana_evaluacion'
]].head(10))
```

La **inconsistencia en los datos categóricos** (sexo, nombres, fechas). Esto dificultaba el análisis porque no se podían agrupar correctamente por categorías.

### ¿Qué técnica de limpieza fue más útil?
La **normalización de texto** (Title Case, mapeo de categorías M/F) fue la más útil porque permitió estandarizar los valores y facilitar el análisis posterior.

### ¿Qué característica nueva aporta más valor al proyecto?
`riesgo_academico` es la más valiosa porque **combina múltiples indicadores** (nota final, probabilidad de deserción) en una sola clasificación que permite identificar estudiantes en riesgo de forma temprana.

Las otras 5 características complementan el análisis:
- `promedio_academico`: rendimiento general
- `porcentaje_asistencia`: nivel de compromiso con clases
- `actividad_lms`: participación en plataforma
- `es_fin_semana`: estacionalidad de las intervenciones
