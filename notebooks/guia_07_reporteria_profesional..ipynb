```python
# IMPORTAR LIBRERÍAS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt              # Gráficos estáticos (estable)
from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo

# Hora de El Salvador (UTC-6)
ZONA_SV = ZoneInfo('America/El_Salvador')
def ahora_sv():
    return datetime.now(ZONA_SV)
from jinja2 import Template
import base64
import json
import warnings
warnings.filterwarnings('ignore')  # Ignorar warnings menores

# CONFIGURAR MATPLOTLIB
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

print(" Librerías importadas")
print(f"   - pandas: {pd.__version__}")
print(f"   - matplotlib: {plt.matplotlib.__version__}")

# CARGAR DATASETS
df = pd.read_csv(BASE / 'datos' / 'procesados' / 'dataset_limpio_Proyecto_EdTech_Learning_Analytics.csv')

df_municipios = pd.read_csv(BASE / 'datos' / 'geo' / 'municipios_alertacampus_guia06_validado.csv')
df_alertas = pd.read_csv(BASE / 'datos' / 'geo' / 'alertas_sinteticas_alertacampus_guia06_validado.csv')
df_alertas['fecha'] = pd.to_datetime(df_alertas['fecha'])

print(f"\nDatasets cargados:")
print(f"   - Estudiantes: {df.shape[0]} filas × {df.shape[1]} columnas")
print(f"   - Municipios: {df_municipios.shape[0]} filas")
print(f"   - Alertas: {df_alertas.shape[0]} filas")

# Solo columnas no personales (gobernanza: sin nombres, teléfono ni correo)
columnas_seguras = ['facultad', 'carrera', 'materia', 'nota_final',
                    'porcentaje_asistencia', 'actividad_lms', 'riesgo_academico']
print(f"\nPrimeras 3 filas del dataset principal (columnas no personales):")
print(df[columnas_seguras].head(3).to_string())
```

```python
# CALCULAR KPIs
df['es_riesgo_alto'] = (df['riesgo_academico'] == 'Alto')

kpis = {
    'total_estudiantes':   df['carnet'].nunique(),
    'promedio_academico':  df['promedio_academico'].mean(),
    'asistencia_prom':     df['porcentaje_asistencia'].mean(),
    'actividad_lms_prom':  df['actividad_lms'].mean(),
    'riesgo_alto_pct':     df['es_riesgo_alto'].mean() * 100,
    'fecha_inicio':        df_alertas['fecha'].min().strftime('%d/%m/%Y'),
    'fecha_fin':           df_alertas['fecha'].max().strftime('%d/%m/%Y'),
    'periodo':             f"{df_alertas['fecha'].min().strftime('%d/%m/%Y')} - {df_alertas['fecha'].max().strftime('%d/%m/%Y')}"
}

print("KPIs DEL PERIODO")
print("=" * 50)
print(f"Total Estudiantes:    {kpis['total_estudiantes']:>10}")
print(f"Promedio Académico:   {kpis['promedio_academico']:>10.2f}")
print(f"Asistencia Promedio:  {kpis['asistencia_prom']:>9.1f}%")
print(f"Actividad LMS Prom.:  {kpis['actividad_lms_prom']:>10.1f}")
print(f"Riesgo Alto:          {kpis['riesgo_alto_pct']:>9.1f}%")
print(f"Periodo:              {kpis['periodo']}")
print("=" * 50)
```

```python
#ANÁLISIS POR FACULTAD
df_categoria = df.groupby('facultad').agg(
    estudiantes=('carnet', 'count'),
    promedio=('promedio_academico', 'mean'),
    asistencia=('porcentaje_asistencia', 'mean'),
    riesgo_alto=('es_riesgo_alto', 'sum')
).reset_index()
df_categoria['pct_riesgo_alto'] = (df_categoria['riesgo_alto'] / df_categoria['estudiantes']) * 100
df_categoria = df_categoria.sort_values('pct_riesgo_alto', ascending=False)

print("ANÁLISIS POR CATEGORÍA (FACULTAD):")
print(df_categoria.round(2).to_string(index=False))

#ANÁLISIS POR NIVEL DE RIESGO
df_canal = df.groupby('riesgo_academico').agg(
    estudiantes=('carnet', 'count'),
    promedio=('promedio_academico', 'mean'),
    asistencia=('porcentaje_asistencia', 'mean'),
    actividad_lms=('actividad_lms', 'mean')
).reset_index()
df_canal = df_canal.sort_values('estudiantes', ascending=False)

print("\nANÁLISIS POR CANAL (NIVEL DE RIESGO):")
print(df_canal.round(2).to_string(index=False))

#ANÁLISIS POR REGIÓN (municipios)
df_region = df_municipios[['municipio', 'zona', 'estudiantes',
                           'promedio_academico', 'pct_riesgo_academico_alto']].copy()
df_region = df_region.sort_values('pct_riesgo_academico_alto', ascending=False)

print("\nANÁLISIS POR REGIÓN (MUNICIPIO):")
print(df_region.round(2).to_string(index=False))

#INTERVENCIONES POR MES (para el gráfico de tendencia)
df_alertas['mes'] = df_alertas['fecha'].dt.month
df_mes = df_alertas.dropna(subset=['fecha']).groupby('mes').agg(
    intervenciones=('id_evento', 'count')
).reset_index()
df_mes['mes'] = df_mes['mes'].astype(int)

print("\nINTERVENCIONES POR MES:")
print(df_mes.to_string(index=False))
```

```python
# ===================================================================
# GRÁFICO 1: Tendencia temporal (líneas)
# ===================================================================
meses = {1: 'Ene', 2: 'Feb', 3: 'Mar', 4: 'Abr', 5: 'May', 6: 'Jun',
         7: 'Jul', 8: 'Ago', 9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dic'}

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(df_mes['mes'], df_mes['intervenciones'],
        marker='o', linewidth=2.5, markersize=10,
        color='#2c7bb6', markerfacecolor='#1a5490')
ax.fill_between(df_mes['mes'], df_mes['intervenciones'], alpha=0.2, color='#2c7bb6')
ax.set_title('Evolución de Intervenciones por Mes', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Mes (2026)', fontsize=11)
ax.set_ylabel('Intervenciones', fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_xticks(df_mes['mes'])
ax.set_xticklabels([meses[m] for m in df_mes['mes']])

# Añadir etiquetas con valores
for mes, valor in zip(df_mes['mes'], df_mes['intervenciones']):
    ax.annotate(f'{valor}', (mes, valor),
                textcoords="offset points", xytext=(0, 10),
                ha='center', fontsize=9, fontweight='bold')

plt.tight_layout()
ruta_tendencia = BASE / 'evidencias' / 'reporte' / 'grafico_tendencia.png'
plt.savefig(ruta_tendencia, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 1: {ruta_tendencia.name}")

# ===================================================================
# GRÁFICO 2: Riesgo alto por categoría (facultad) (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(10, 6))
colores = plt.cm.Blues(np.linspace(0.4, 0.9, len(df_categoria)))
barras = ax.bar(df_categoria['facultad'], df_categoria['pct_riesgo_alto'],
                color=colores, edgecolor='#1a5490', linewidth=1.5)
ax.set_title('Estudiantes en Riesgo Alto por Facultad (%)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Facultad', fontsize=11)
ax.set_ylabel('Riesgo alto (%)', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')
ax.tick_params(axis='x', rotation=30)
for etiqueta in ax.get_xticklabels():
    etiqueta.set_ha('right')

for barra, valor in zip(barras, df_categoria['pct_riesgo_alto']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'{valor:.1f}%', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_categoria = BASE / 'evidencias' / 'reporte' / 'grafico_categorias.png'
plt.savefig(ruta_categoria, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 2: {ruta_categoria.name}")

# ===================================================================
# GRÁFICO 3: Estudiantes por canal (nivel de riesgo) (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(10, 6))
colores = ['#e74c3c', '#f39c12', '#2ecc71']
barras = ax.bar(df_canal['riesgo_academico'], df_canal['estudiantes'],
                color=colores, edgecolor='#2c3e50', linewidth=1.5)
ax.set_title('Estudiantes por Nivel de Riesgo', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Nivel de riesgo', fontsize=11)
ax.set_ylabel('Estudiantes', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')

for barra, valor in zip(barras, df_canal['estudiantes']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'{valor}', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_canal = BASE / 'evidencias' / 'reporte' / 'grafico_canal.png'
plt.savefig(ruta_canal, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 3: {ruta_canal.name}")

# ===================================================================
# GRÁFICO 4: Riesgo alto por región (municipio) (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(12, 6))
colores = plt.cm.Greens(np.linspace(0.4, 0.9, len(df_region)))
barras = ax.bar(df_region['municipio'], df_region['pct_riesgo_academico_alto'],
                color=colores, edgecolor='#1a5490', linewidth=1.5)
ax.set_title('Estudiantes en Riesgo Alto por Municipio (Análisis Geoespacial)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Municipio', fontsize=11)
ax.set_ylabel('Riesgo alto (%)', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')
ax.tick_params(axis='x', rotation=45)
for etiqueta in ax.get_xticklabels():
    etiqueta.set_ha('right')

for barra, valor in zip(barras, df_region['pct_riesgo_academico_alto']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'{valor:.1f}%', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_region = BASE / 'evidencias' / 'reporte' / 'grafico_region.png'
plt.savefig(ruta_region, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 4: {ruta_region.name}")

print("\n¡Los 4 gráficos se generaron correctamente!")
```

```python
# GENERAR DOCX CON PYTHON-DOCX
from docx import Document
from docx.shared import Inches, Pt, Cm
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

doc = Document()

# Formato de los lineamientos: Arial 11, interlineado 1.15, márgenes 2.5 cm
for s in doc.sections:
    s.left_margin = s.right_margin = s.top_margin = s.bottom_margin = Cm(2.5)
for nombre_estilo in ['Normal', 'Title', 'Heading 1', 'Heading 2', 'Heading 3', 'List Bullet']:
    est = doc.styles[nombre_estilo]
    est.font.name = 'Arial'
    est.element.rPr.rFonts.set(qn('w:eastAsia'), 'Arial')
doc.styles['Normal'].font.size = Pt(11)
doc.styles['Normal'].paragraph_format.line_spacing = 1.15

# Encabezado y número de página (abajo a la derecha)
sec = doc.sections[0]
sec.header.paragraphs[0].text = f'{NOMBRE_PROYECTO} - Informe gerencial (Guía 7) - Avance 2 - Sección {SECCION}'
pie = sec.footer.paragraphs[0]
pie.alignment = WD_ALIGN_PARAGRAPH.RIGHT
pie.add_run('Página ')
_run = pie.add_run()
_f1 = OxmlElement('w:fldChar'); _f1.set(qn('w:fldCharType'), 'begin')
_it = OxmlElement('w:instrText'); _it.set(qn('xml:space'), 'preserve'); _it.text = 'PAGE'
_f2 = OxmlElement('w:fldChar'); _f2.set(qn('w:fldCharType'), 'end')
_run._r.append(_f1); _run._r.append(_it); _run._r.append(_f2)

def tabla(headers, filas):
    t = doc.add_table(rows=1, cols=len(headers))
    t.style = 'Light Grid Accent 1'
    for i, h in enumerate(headers):
        t.rows[0].cells[i].text = h
    for fila in filas:
        celdas = t.add_row().cells
        for i, v in enumerate(fila):
            celdas[i].text = str(v)
    return t

def figura(ruta, leyenda, ancho=6):
    doc.add_picture(str(ruta), width=Inches(ancho))
    doc.paragraphs[-1].alignment = WD_ALIGN_PARAGRAPH.CENTER
    p = doc.add_paragraph(leyenda)
    p.runs[0].italic = True
    p.runs[0].font.size = Pt(9)

# Portada
titulo = doc.add_heading('INFORME GERENCIAL DE RENDIMIENTO ACADÉMICO', 0)
titulo.alignment = WD_ALIGN_PARAGRAPH.CENTER
subtitulo = doc.add_heading(NOMBRE_PROYECTO, level=1)
subtitulo.alignment = WD_ALIGN_PARAGRAPH.CENTER
doc.add_paragraph(f'Periodo: {kpis["periodo"]}')
doc.add_paragraph(f'Fecha de generación: {ahora_sv().strftime("%d/%m/%Y %H:%M")}')
doc.add_paragraph('Versión: 1.0')
doc.add_paragraph(f'Equipo: {datos_template["equipo"]} - Sección {SECCION}')
doc.add_page_break()

# 1. Resumen ejecutivo
doc.add_heading('1. Resumen Ejecutivo', level=2)
doc.add_paragraph(
    f'El presente informe analiza el rendimiento académico y el riesgo de deserción de {kpis["total_estudiantes"]} '
    f'estudiantes de la sede central durante el periodo {kpis["periodo"]}. Su propósito es apoyar la decisión de a qué '
    f'estudiantes y zonas dirigir tutorías y acompañamiento.')
doc.add_paragraph(f'Periodo analizado: {kpis["periodo"]}')
doc.add_heading('Indicadores Principales', level=3)
tabla(['KPI', 'Valor', 'Unidad', 'Periodo', 'Comparación'],
      [[f['kpi'], f['valor'], f['unidad'], f['periodo'], f['comparacion']] for f in datos_template['filas_kpi']])
doc.add_heading('Hallazgos clave', level=3)
for h in datos_template['hallazgos']:
    doc.add_paragraph(h['titulo'], style='List Number')
doc.add_heading('Recomendaciones priorizadas', level=3)
for r_ in datos_template['recomendaciones']:
    doc.add_paragraph(f'{r_["prioridad"]}: {r_["accion"]}', style='List Bullet')
doc.add_heading('Riesgos y limitaciones', level=3)
for l in datos_template['limitaciones'][:2]:
    doc.add_paragraph(l['limitacion'], style='List Bullet')
doc.add_page_break()

# 2. Contexto y metodología
doc.add_heading('2. Contexto y Metodología', level=2)
doc.add_paragraph('Fuentes utilizadas: Dataset limpio (Guía 2), Dashboard Power BI (Guía 5), Análisis geoespacial ArcGIS (Guía 6).')
doc.add_paragraph('Proceso: Integración → Limpieza → Procesamiento → Gobernanza → Visualización → Análisis territorial.')
doc.add_paragraph(f'Filtros aplicados: Periodo {kpis["periodo"]}, todas las facultades, todos los niveles de riesgo, todos los municipios.')
doc.add_paragraph('Definición de riesgo: Alto si la nota final es menor a 6 o la probabilidad de deserción es mayor a 0.7; '
                  'Medio si la nota es menor a 7 o la probabilidad es mayor a 0.4; Bajo en los demás casos.')
doc.add_paragraph('Privacidad: los datos son sintéticos y el informe solo presenta datos agregados, sin nombres, contactos ni identificadores personales.')

# 3. Resultados generales
doc.add_heading('3. Resultados Generales', level=2)
doc.add_heading('3.1 Evolución de Intervenciones', level=3)
figura(ruta_tendencia, 'Figura 1. Intervenciones por mes (Python/matplotlib).')
doc.add_heading('3.2 Riesgo Alto por Facultad', level=3)
figura(ruta_categoria, 'Figura 2. Porcentaje de estudiantes en riesgo alto por facultad.')
doc.add_heading('3.3 Estudiantes por Nivel de Riesgo', level=3)
figura(ruta_canal, 'Figura 3. Estudiantes por nivel de riesgo.')
doc.add_heading('Visualizaciones integradas', level=3)
tabla(['N.º', 'Visual', 'Fuente', 'Mensaje', 'Pregunta que responde'],
      [[v['n'], v['visual'], v['fuente'], v['mensaje'], v['pregunta']] for v in datos_template['filas_visuales']])
if lista_pbi:
    doc.add_heading('3.4 Dashboard Interactivo (Power BI)', level=3)
    for d in lista_pbi:
        figura(d['ruta'], f'{d["titulo"]}. Dashboard de la Guía 5, con todos los filtros activos.')

# 4. Análisis geoespacial
doc.add_heading('4. Análisis Geoespacial', level=2)
if not capas_pendientes:
    doc.add_heading('4.1 Capas configuradas (ArcGIS)', level=3)
    tabla(['Capa', 'Tipo', 'Fuente', 'Variable principal', 'Estilo', 'Propósito', 'Privacidad'],
          [[c['capa'], c['tipo'], c['fuente'], c['variable'], c['estilo'], c['proposito'], c['privacidad']] for c in CAPAS_ARCGIS])
if lista_mapas:
    doc.add_heading('4.2 Mapas', level=3)
    for i, m in enumerate(lista_mapas[:MAPAS_EN_CUERPO], 1):
        figura(m['ruta'], f'Figura {i}. {m["titulo"]}. {m["descripcion"]} Fuente: Guía 6 (ArcGIS). Datos sintéticos y agregados.')
doc.add_heading('4.3 Riesgo alto por municipio', level=3)
figura(ruta_region, 'Riesgo alto por municipio (Python/matplotlib).')
doc.add_paragraph(
    f'Patrón observado: {datos_template["region_top"]} presenta el mayor porcentaje de estudiantes en riesgo alto '
    f'({datos_template["pct_region_top"]:.1f}%) y {datos_template["region_baja"]} el menor ({datos_template["pct_region_baja"]:.1f}%). '
    f'La zona rural tiene {datos_template["pct_rural"]:.1f}% y la zona urbana {datos_template["pct_urbana"]:.1f}%.')
doc.add_paragraph('Lectura responsable: las diferencias son moderadas y no concluyentes. El mapa no demuestra causalidad.')
doc.add_heading('4.4 Hallazgos territoriales', level=3)
for i, g in enumerate(datos_template['hallazgos_geo'], 1):
    doc.add_heading(f'{i}. {g["titulo"]}', level=3)
    doc.add_paragraph(f'Evidencia: {g["evidencia"]}')
    doc.add_paragraph(f'Interpretación: {g["interpretacion"]}')
    doc.add_paragraph(f'Acción sugerida: {g["accion"]}')
    doc.add_paragraph(f'Limitación: {g["limitacion"]}')

# 5. Hallazgos principales
doc.add_heading('5. Hallazgos Principales', level=2)
for i, h in enumerate(datos_template['hallazgos'], 1):
    doc.add_heading(f'{i}. {h["titulo"]}', level=3)
    doc.add_paragraph(f'Evidencia: {h["evidencia"]}')
    doc.add_paragraph(f'Interpretación: {h["interpretacion"]}')
    doc.add_paragraph(f'Impacto: {h["impacto"]}')
    doc.add_paragraph(f'Decisión sugerida: {h["decision"]}')
    doc.add_paragraph(f'Limitación: {h["limitacion"]}')

# 6. Recomendaciones (7 columnas, como en los lineamientos)
doc.add_heading('6. Recomendaciones', level=2)
tabla(['Hallazgo', 'Acción recomendada', 'Prioridad', 'Responsable', 'Plazo', 'KPI de seguimiento', 'Riesgo'],
      [[r_['hallazgo'], r_['accion'], r_['prioridad'], r_['responsable'], r_['plazo'], r_['kpi'], r_['riesgo']]
       for r_ in datos_template['recomendaciones']])

# 7. Limitaciones (tabla limitación / impacto / mitigación)
doc.add_heading('7. Limitaciones del Análisis', level=2)
tabla(['Limitación', 'Impacto', 'Mitigación'],
      [[l['limitacion'], l['impacto'], l['mitigacion']] for l in datos_template['limitaciones']])

# 8. Conclusiones
doc.add_heading('8. Conclusiones', level=2)
doc.add_paragraph(datos_template['conclusion'])

# Anexo A: capturas adicionales del mapa
if len(lista_mapas) > MAPAS_EN_CUERPO:
    doc.add_page_break()
    doc.add_heading('Anexo A. Capturas adicionales del mapa (ArcGIS)', level=2)
    for i, m in enumerate(lista_mapas[MAPAS_EN_CUERPO:], 1):
        figura(m['ruta'], f'Figura A{i}. {m["titulo"]}. {m["descripcion"]} Fuente: Guía 6 (ArcGIS). Datos sintéticos y agregados.')

# Guardar
ruta_docx = BASE / 'exportaciones' / f'informe_{NOMBRE_ARCHIVO}_v1.0.docx'
doc.save(str(ruta_docx))
print(f"DOCX generado: {ruta_docx.name}")
print(f"Tamaño: {ruta_docx.stat().st_size / 1024:.1f} KB")

files.download(str(ruta_docx))
print("DOCX descargado")
```

```python
# MATRICES EN EXCEL (entregables)
#   - matriz_recomendaciones_guia07.xlsx
#   - matriz_trazabilidad_guia07.xlsx
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

def guardar_excel(df_, ruta, hoja):
    with pd.ExcelWriter(ruta, engine='openpyxl') as w:
        df_.to_excel(w, index=False, sheet_name=hoja)
        ws = w.sheets[hoja]
        for celda in ws[1]:
            celda.font = Font(bold=True, color='FFFFFF')
            celda.fill = PatternFill('solid', fgColor='2C7BB6')
            celda.alignment = Alignment(wrap_text=True, vertical='center')
        for col in ws.columns:
            letra = get_column_letter(col[0].column)
            ancho = max(len(str(c.value)) if c.value is not None else 0 for c in col)
            ws.column_dimensions[letra].width = min(max(14, ancho * 0.6), 60)
            for c in col[1:]:
                c.alignment = Alignment(wrap_text=True, vertical='top')
        ws.freeze_panes = 'A2'

carpeta_docs = BASE / 'documentos'
carpeta_docs.mkdir(parents=True, exist_ok=True)

# --- Matriz de recomendaciones
hall = datos_template['hallazgos']
df_rec = pd.DataFrame([{
    'Hallazgo': f"H{r_['hallazgo']}: {hall[r_['hallazgo'] - 1]['titulo']}",
    'Acción recomendada': r_['accion'],
    'Prioridad': r_['prioridad'],
    'Responsable': r_['responsable'],
    'Plazo': r_['plazo'],
    'KPI de seguimiento': r_['kpi'],
    'Riesgo': r_['riesgo'],
} for r_ in datos_template['recomendaciones']])
ruta_rec_xlsx = carpeta_docs / 'matriz_recomendaciones_guia07.xlsx'
guardar_excel(df_rec, ruta_rec_xlsx, 'Recomendaciones')

# --- Matriz de trazabilidad
FUENTE_DF  = 'dataset_limpio_Proyecto_EdTech_Learning_Analytics.csv (Guía 2)'
FUENTE_MUN = 'municipios_alertacampus_guia06_validado.csv (Guía 6)'
fuentes_hallazgos = [FUENTE_DF, FUENTE_DF, FUENTE_MUN]   # hallazgo 1, 2 y 3 del informe
filas_traz = []
for i, h in enumerate(datos_template['hallazgos']):
    filas_traz.append({
        'N.º': f'H{i + 1}', 'Hallazgo': h['titulo'], 'Evidencia': h['evidencia'],
        'Fuente': fuentes_hallazgos[i] if i < len(fuentes_hallazgos) else FUENTE_DF,
        'Filtro / Periodo': f"{kpis['periodo']}; todos los filtros",
        'Limitación': h['limitacion'], 'Decisión posible': h['decision']})
for i, g in enumerate(datos_template['hallazgos_geo']):
    filas_traz.append({
        'N.º': f'G{i + 1}', 'Hallazgo': g['titulo'], 'Evidencia': g['evidencia'],
        'Fuente': FUENTE_MUN, 'Filtro / Periodo': f"{kpis['periodo']}; todos los municipios",
        'Limitación': g['limitacion'], 'Decisión posible': g['accion']})
df_traz = pd.DataFrame(filas_traz)
ruta_traz_xlsx = carpeta_docs / 'matriz_trazabilidad_guia07.xlsx'
guardar_excel(df_traz, ruta_traz_xlsx, 'Trazabilidad')

print(f"{ruta_rec_xlsx.name}: {len(df_rec)} recomendaciones")
print(f"{ruta_traz_xlsx.name}: {len(df_traz)} hallazgos trazados")

files.download(str(ruta_rec_xlsx))
files.download(str(ruta_traz_xlsx))
```
