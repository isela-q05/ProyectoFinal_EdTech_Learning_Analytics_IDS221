```python
# ==========================================
# PROYECTO EDTECH LEARNING ANALYTICS - PARTE 4
#GOBERNANZA Y ETICA DE DATOS
# ==========================================
#IMPORTACION DE LIBRERIAS
import pandas as pd
import numpy as np
import hashlib
from pathlib import Path

print("Librerias importadas --------------")
```

```python
#CREAR CARPETAS PARA GOBERNANZA

BASE_DIR = Path("/content/")
DATA_GOV = BASE_DIR/"datos"/"gobernanza"
DOCS_DIR = BASE_DIR/"documentos"

DATA_GOV.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

print(f"Datos gobernanza: {DATA_GOV}")
print(f"Documentos: {DOCS_DIR}")
```

```python
ruta_alumnado = BASE_DIR / "datos" / "gobernanza" / "dataset_limpio_Proyecto_EdTech_Learning_Analytics.csv"

df_alumnadoUES = pd.read_csv(ruta_alumnado)

print(f"Documento cargado correctamente desde: {ruta_alumnado}")
df_alumnadoUES.head()
```

```python
#CLASIFICACION DE DATOS
#NIVELES: PUBLICOS, PRIVADOS, CONFIDENCIAL, RESTRINGIDO

clasificacion = pd.DataFrame([
    ["carnet", "Identificador interno", "Confidencial", "Vincular registros", "Seudonimizar con SHA-256"],
    ["nombres", "Identificador directo", "Confidencial", "Contacto institucional", "Excluir de vista analítica"],
    ["apellidos", "Identificador directo", "Confidencial", "Contacto institucional", "Excluir de vista analítica"],
    ["telefono", "Contacto", "Confidencial", "Urgencias / Avisos", "Excluir o enmascarar"],
    ["correo", "Contacto", "Confidencial", "Notificaciones institucionales", "Excluir o enmascarar"],
    ["fecha_nacimiento", "Demográfico directo", "Confidencial", "Cálculo de edad", "Excluir tras calcular edad"],
    ["edad", "Demográfico", "Interno", "Segmentación académica", "Generalizar por rango de edad"],
    ["sexo", "Demográfico", "Interno", "Estadística de equidad", "Conservar para reportes agrupados"],
    ["departamento", "Ubicación", "Interno", "Análisis territorial", "Revisar granularidad regional"],
    ["zona", "Ubicación", "Interno", "Análisis de entorno (Urbana/Rural)", "Conservar"],
    ["facultad", "Académico", "Público", "Estadística general", "Conservar"],
    ["carrera", "Académico", "Público", "Reportes institucionales", "Conservar"],
    ["anio_ingreso", "Académico", "Público", "Cohorte estudiantil", "Conservar"],
    ["ciclo_actual", "Académico", "Público", "Avance de plan de estudio", "Conservar"],
    ["turno", "Académico", "Público", "Gestión de horarios", "Conservar"],
    ["acceso_internet", "Socioeconómico", "Restringido", "Diagnóstico de brecha digital", "Acceso condicionado a Bienestar Estudiantil"],
    ["trabaja", "Socioeconómico", "Restringido", "Evaluación de condición laboral", "Acceso restringido"],
    ["horas_trabajo", "Socioeconómico", "Restringido", "Medición de carga horaria extra", "Acceso restringido"],
    ["tipo_beca", "Socioeconómico", "Restringido", "Asignación de apoyos financieros", "Acceso restringido"]
],
columns=["campo", "tipo", "clasificacion", "finalidad", "control"])

print("Matriz de clasificación -------------------------")
print(clasificacion)
```

```python
#GUARDAR CLASIFICACION
ruta_clasificacion = DOCS_DIR/"clasificacion_datos_Proyecto_EdTech_Learning_Analytics.xlsx"

clasificacion.to_excel(ruta_clasificacion, index=False)

print(f"Clasificacion guardada {ruta_clasificacion}")
```

```python
#SEUDONIMIZACION (SERIA REMPLACAR IDENTIDAD)

import hashlib

SALT = "Proyecto-EdTech-Learning-Analytics-datos-sinteticos"

def tokenizar(valor):
    """ Convierte el valor en un codigo unico de 12 caracteres."""
    texto = f"{SALT}|{valor}".encode('utf-8')
    return hashlib.sha256(texto).hexdigest()[:12]

df_alumnadoUES["estudiante_token"] = df_alumnadoUES["carnet"].astype(str).apply(tokenizar)

print("Seudonimización aplicada: ------------")
print(df_alumnadoUES[["carnet", "estudiante_token"]])
```

```python
# ENMASCARAMIENTO DEL CORREO (NO SE MUESTRA EL CORREO COMPLETO SOLO UNA PARTE)
def enmascarar_correo(correo):
    """Muestra solo la primera letra del correo electronico y el dominio completo"""
    if pd.isna(correo):
        return correo
    usuario, dominio = str(correo).split('@', 1)
    visible = usuario[:1]
    return f"{visible}***@{dominio}"

df_alumnadoUES["correo_mask"] = df_alumnadoUES["correo_institucional"].apply(enmascarar_correo)

print("Correo enmascarado--------------------")
print(df_alumnadoUES[["correo_institucional", "correo_mask"]])
```

```python
#ENMASCARMIENTO PARA EL TELEFONO

df_alumnadoUES["telefono_mask"] = df_alumnadoUES["telefono"].astype(str).apply(lambda x: "****" +x[-4:])

print("Telefono enmascarado -------------")
print(df_alumnadoUES[["telefono", "telefono_mask"]])
```

```python
#GENERALIZACION DE EDAD

bins = [0, 24, 34, 44, 54, 64, 200]
labels = ["<=24", "25-34", "35-44", "45-54", "55-64", "65+"]

df_alumnadoUES["rango_edad"] = pd.cut(df_alumnadoUES["edad"], bins=bins, labels=labels)

print("Edad generalizada: -------------------------")
print(df_alumnadoUES[["edad", "rango_edad"]])
```

```python
#CREAR VISTA ANALÍTICA PROTEGIDA
#PRINCIPIP DE MINIZACION

vista_analitica = df_alumnadoUES[[
    "estudiante_token",
    "rango_edad",
    "telefono_mask",
    "correo_mask",
    "municipio_residencia",
    "facultad",
    "carrera"
]].copy()

print("VISTA ANALITICA PROTEGIDA(SIN IDENTIFICADORES DIRECTOS)")
print(vista_analitica)
```

```python
#GUIA VISTA PROTEGIDA

ruta_publicable = DATA_GOV/"datase_publicate_Proyecto_EdTech_Learning_Analytics.csv"

vista_analitica.to_csv(ruta_publicable, index=False, encoding='utf-8')

print(f"***VISTA ANALITICA PROTEGIDA GUARDADA:{ruta_publicable}***")
```

```python
#VALIDACION AUTOMATICA

identificadores_directos = {"nombres", "apellidos","edad", "correo_institucional", "telefono", "carnet"}
expuestos = identificadores_directos.intersection(vista_analitica.columns)

print("Verificando identificadores directos: --------------------")
print(f"Indentificadores directos expuestos:{expuestos}")

if len(expuestos) == 0 :
  print("Ningun identificador directo expuestos!!!!!")
else:
  print("Alerta: Identificadores directos expuestos!!!!!!!")
```

```python
#MATRIZ DE RIESGOS - IDENTICAR, PRIORIZAR Y CONTROLAR RIESGOS
#ESCALA: PROBABILIDAD E IMPACTO DEL 1 AL 5
#1 = MUY BAJO, 5= MUY ALTO
#PUNTAJE = PROBABILIDAD x IMPACTO

riesgos = pd.DataFrame([
    ["Exposición de identificadores directos", 4, 5, "Eliminar o enmascarar campos", "Líder de datos", "Dataset protegido"],
    ["Reidentificación por combinación de atributos", 3, 4, "Generalizar y revisar granularidad", "Analista", "Análisis de grupos"],
    ["Acceso excesivo al dataset crudo", 3, 5, "RBAC y mínimo privilegio", "Administrador", "Matriz de acceso"],
    ["Retención indefinida de copias", 3, 3, "Regla de retención", "Líder de proyecto", "Bitácora de eliminación"],
    ["Uso del dato para finalidad no definida", 2, 5, "Registrar propósito", "Propietario del dato", "Ficha de finalidad"],
    ["Conclusiones sesgadas por segmentación", 3, 4, "Revisión ética", "Equipo analítico", "Reflexión y pruebas"]
],
columns=["riesgo", "probabilidad", "impacto", "control", "responsable", "evidencia"])

riesgos["puntaje"] = riesgos["probabilidad"] * riesgos["impacto"]
riesgos["nivel"] = riesgos["puntaje"].apply(
    lambda x: "Crítico" if x >= 16 else "Alto" if x >= 10 else "Moderado" if x >= 5 else "Bajo"
)

print("MATRIZ DE RIESGOS:")
print(riesgos.sort_values("puntaje", ascending=False))
```

```python
#GUARDAR LA MATRIZ DE RIESGOS

ruta_riesgos = DOCS_DIR/"matriz_riesgos_Proyecto_EdTech_Learning_Analytics.xlsx"
riesgos.to_excel(ruta_riesgos, index=False)

print(f"Matriz de riesgos guardada --------------{ruta_riesgos}")
```
